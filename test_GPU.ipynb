Wenn Sie dieses Notebook verwenden, sollten Sie bereits im Moodle-Kurs für den EFI-GPU-Server eingetragen sein. Damit sollten Sie entweder bereits für den GPU-Server freigeschaltet sein oder dies bald erfolgen. Dieses Notebook erklärt Schritt für Schritt, wie Sie den Server mit Visual Studio Code (VSC) und einem Jupyter-Notebook verwenden können.

Geben Sie eine der folgenden URLs in einem Browser ein und loggen Sie sich mit Ihren VirtuOhm-Zugangsdaten ein: 
* Für reguläre Lehrveranstaltungen:   https://www.efi.th-nuernberg.de/ki0/
* Für Abschlussarbeiten und Projekte: https://www.efi.th-nuernberg.de/ki1/

Sollten Sie sich nicht einloggen können, sprechen Sie bitte Ihre/n betreuende/n Dozent/in an.

Sollten Ihnen nun mehrere Auswahl-Optionen angezeigt werden, fragen Sie bitte Ihre/n Dozent/in, welche Sie verwenden sollen.

Nach Auswahl einer Option sollte sich nun ein Server-Fenster öffnen.

Um nun den eben gestarteten Server mit VSC zu verwenden, muss der Server in VSC hinterlegt werden. Wenn Sie ein Notebook in VSC geöffnet haben, sollte oben rechts die Option "Select Kernel" verfügbar sein. Evtl. haben Sie bereits einen Kernel ausgewählt. In diesem Fall wird der Kernel (z.B. "Python 3.9") angezeigt. Klicken Sie nun darauf. Es öffnet sich nun ein Auswahlmenü. Wählen Sie hier "Select another Kernel", danach "Existing JupyterHub Server" und anschließend "Enter the URL of the running JupyterHub Server". 

Sollte "Existing JupyterHub Server" nicht verfügbar sein, wählen Sie "Existing Jupyter Server". Geben Sie nun die URL von oben ein mit der Erweiterung `user/<Benutzername>` (für `<Benutzername>` geben Sie bitte Ihren VirtuOhm-Benutzernamen an), also z.B. https://www.efi.th-nuernberg.de/ki0/user/mustermannma1234. Falls die Option "Existing JupyterHub Server" nicht verfügbar war, sollte VSC Sie nun dazu auffordern ein entsprechendes Plugin zu installieren. Tun Sie dies, starten Sie VSC neu und wiederholen Sie die Schritte bis hierher.

VSC fragt Sie nun zunächst nach Ihrem Benutzernamen und anschließend nach Ihrem Passwort. Geben Sie hier wieder Ihre VirtuOhm-Zugangsdaten ein. Abschließend können Sie einen Namen für die Konfiguration vergeben. Unter diesem Namen können Sie auch mit anderen Notebooks den Server wieder unter "Existing JupyterHub Server" auswählen, ohne die Konfiguration erneut durchzuführen.
